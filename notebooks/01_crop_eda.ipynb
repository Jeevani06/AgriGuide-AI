```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df = pd.read_csv("../data/raw/crop_recommendation.csv")

df.head(); df.info(); df.describe()      # one cell each
df.isnull().sum()                        # missing values
df["label"].value_counts()               # class balance (22 crops, ~100 each)

df.hist(figsize=(12, 8))                                     # feature distributions
sns.heatmap(df.drop(columns="label").corr(), annot=True)     # correlations
sns.boxplot(data=df, x="label", y="rainfall"); plt.xticks(rotation=90)
```
