```python
import pandas as pd, joblib
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

y = pd.read_csv("../data/raw/crop_yield.csv").dropna()
y["Crop"] = y["Crop"].str.strip().str.lower()
y = y[["Crop", "Annual_Rainfall", "Fertilizer", "Pesticide", "Yield"]]
y.to_csv("../data/processed/crop_yield_clean.csv", index=False)

crops = sorted(y["Crop"].unique())
X = pd.DataFrame({"crop": y["Crop"].map({c: i for i, c in enumerate(crops)}),
                  "rain": y["Annual_Rainfall"], "fert": y["Fertilizer"], "pest": y["Pesticide"]})
Xtr, Xte, ytr, yte = train_test_split(X, y["Yield"], test_size=0.2, random_state=1)

for name, m in {"Linear": LinearRegression(), "Random Forest": RandomForestRegressor(200, random_state=1)}.items():
    m.fit(Xtr, ytr); p = m.predict(Xte)
    print(name, "R2:", r2_score(yte, p), "MAE:", mean_absolute_error(yte, p))

rf = RandomForestRegressor(200, random_state=1).fit(Xtr, ytr)
joblib.dump({"model": rf, "crops": crops}, "../models/yield_prediction_model.pkl")
```
